#### 2.1.2 Customer Type per Station [10 points]

There are two kinds of long-term passes: Indego30 and Indego365. All other passholder types are short-term.

For each station from **2.1.1** (stations with >200 trips), calculate:

  - **Average trip duration** in minutes
  - **Count of member trips** vs **casual trips**
  - **Order by** `avg_duration` DESC, then `station_name` ASC.

**Hint:** Use `COUNT(CASE WHEN ... THEN 1 END)` to count by rider type.

Save your query as `query_station_rider_type` and result as `station_rider_type_df`.**Final Schema:**

> station_id | station_name | avg_duration | long_passholders | short_passholders

````python
eligible_station_ids = (
    trips_df.groupby("start_station_id").size().loc[lambda counts: counts > 200].index
)

station_rider_type_df = (
    trips_df[trips_df["start_station_id"].isin(eligible_station_ids)]
    .merge(
        stations_df[["station_id", "station_name"]],
        left_on="start_station_id",
        right_on="station_id",
        how="inner"
    )
    .groupby(["station_id", "station_name"], as_index=False)
    .agg(
        avg_duration=("duration", "mean"),
        long_passholders=("passholder_type", lambda x: x.isin(["Indego30", "Indego365"]).sum()),
        short_passholders=("passholder_type", lambda x: (~x.isin(["Indego30", "Indego365"])).sum())
    )
    .sort_values(["avg_duration", "station_name"], ascending=[False, True])
    .reset_index(drop=True)
)
````

````python
query_station_rider_type = """
SELECT
    s.station_id,
    s.station_name,
    AVG(t.duration) AS avg_duration,
    COUNT(CASE WHEN t.passholder_type IN ('Indego30', 'Indego365') THEN 1 END) AS long_passholders,
    COUNT(CASE WHEN t.passholder_type NOT IN ('Indego30', 'Indego365') THEN 1 END) AS short_passholders
FROM trips_df AS t
INNER JOIN stations_df AS s
    ON t.start_station_id = s.station_id
GROUP BY
    s.station_id,
    s.station_name
HAVING COUNT(*) > 200
ORDER BY avg_duration DESC, station_name ASC
"""
station_rider_type_sql_df = duckdb.sql(query_station_rider_type).df()
````

````python
trips_per_station_df = (
    trips_df.merge(
        stations_df[["station_id", "neighborhood"]],
        left_on="start_station_id",
        right_on="station_id",
        how="inner"
    )
    .groupby("neighborhood", as_index=False)
    .agg(
        total_trips=("ride_id", "size"),
        unique_stations=("station_id", "nunique")
    )
)
trips_per_station_df["trips_per_station"] = (
    trips_per_station_df["total_trips"] / trips_per_station_df["unique_stations"]
).round(1)
trips_per_station_df = (
    trips_per_station_df
    .sort_values("trips_per_station", ascending=False)
    .reset_index(drop=True)
)
````

````python
query_trips_per_station = """
SELECT
    s.neighborhood,
    COUNT(*) AS total_trips,
    COUNT(DISTINCT s.station_id) AS unique_stations,
    ROUND(COUNT(*) * 1.0 / COUNT(DISTINCT s.station_id), 1) AS trips_per_station
FROM trips_df AS t
INNER JOIN stations_df AS s
    ON t.start_station_id = s.station_id
GROUP BY s.neighborhood
ORDER BY trips_per_station DESC
"""
trips_per_station_sql_df = duckdb.sql(query_trips_per_station).df()
````

Analyze daily ridership trends with smoothing to identify patterns.

**Hint:** Use window function `AVG(...) OVER (ORDER BY ... ROWS BETWEEN 2 PRECEDING AND CURRENT ROW)`.

Save your query as `query_daily_rolling` and result as `daily_rolling_df`.

**Final Schema:**
> date | total_trips | rolling_avg_3day

````python
query_daily_rolling = """
WITH daily AS (SELECT CAST(started_at AS DATE) AS date, COUNT(*) AS total_trips FROM trips_df GROUP BY date)
SELECT date, total_trips, ROUND(AVG(total_trips) OVER (ORDER BY date ROWS BETWEEN 2 PRECEDING AND CURRENT ROW)) AS rolling_avg_3day
FROM daily ORDER BY date ASC
"""
daily_rolling_df = duckdb.sql(query_daily_rolling).df()
````

Save your query as `query_top_stations` and result as `top_stations_df`.

````python
query_top_stations = """
WITH counts AS (SELECT s.station_id, s.station_name, s.neighborhood, COUNT(t.ride_id) AS total_trips,
 ROW_NUMBER() OVER (PARTITION BY s.neighborhood ORDER BY COUNT(t.ride_id) DESC, s.station_name ASC) AS rn,
 COUNT(*) OVER (PARTITION BY s.neighborhood) AS station_count
 FROM stations_df s INNER JOIN trips_df t ON t.start_station_id=s.station_id
 GROUP BY s.station_id,s.station_name,s.neighborhood)
SELECT station_id,station_name,neighborhood,total_trips FROM counts WHERE station_count >= 3 AND rn <= 3
ORDER BY neighborhood ASC,total_trips DESC
"""
top_stations_df = duckdb.sql(query_top_stations).df()
````

Design an ER diagram that captures the relationships above. Be sure to capture all entities, relationships, and cardinalities.

````python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, Polygon
fig, ax = plt.subplots(figsize=(12, 8))
ax.set_xlim(0, 12); ax.set_ylim(0, 8); ax.axis("off")
def entity(x, y, title, attributes):
    ax.add_patch(Rectangle((x, y), 2.5, 1.5, facecolor="#dceeff", edgecolor="black", linewidth=2))
    ax.text(x + 1.25, y + 1.12, title, ha="center", va="center", fontweight="bold", fontsize=12)
    ax.text(x + 1.25, y + .55, attributes, ha="center", va="center", fontsize=10)
def relationship(cx, cy, label):
    ax.add_patch(Polygon([(cx, cy+.55), (cx+.9, cy), (cx, cy-.55), (cx-.9, cy)], closed=True, facecolor="#fff2cc", edgecolor="black", linewidth=2))
    ax.text(cx, cy, label, ha="center", va="center", fontsize=10)
entity(.5, 5.5, "STUDENT", "student_id (PK)\ndemographic_data")
entity(9, 5.5, "COURSE", "course_no (PK)\ndescription")
entity(9, 1, "FACULTY", "faculty_id (PK)\ndemographic_data")
entity(.5, 1, "ADMIN", "admin_id (PK)\ncontact_info")
relationship(6, 6.25, "ENROLLS")
ax.plot([3, 5.1], [6.25, 6.25], color="black"); ax.plot([6.9, 9], [6.25, 6.25], color="black")
ax.text(3.25, 6.45, "M", fontweight="bold"); ax.text(8.65, 6.45, "N", fontweight="bold")
relationship(10.25, 4.0, "TEACHES")
ax.plot([10.25, 10.25], [5.5, 4.55], color="black"); ax.plot([10.25, 10.25], [3.45, 2.5], color="black")
ax.text(10.45, 5.25, "M", fontweight="bold"); ax.text(10.45, 2.65, "N", fontweight="bold")
relationship(6, 1.75, "ASSIGNED_TO")
ax.plot([3, 5.1], [1.75, 1.75], color="black"); ax.plot([6.9, 9], [1.75, 1.75], color="black")
ax.text(3.25, 1.95, "1 (one admin)", fontweight="bold"); ax.text(7.7, 1.95, "N (many faculty)", fontweight="bold")
ax.set_title("University Enrollment ER Diagram", fontsize=16, fontweight="bold")
fig.tight_layout(); fig.savefig("your_er_diagram.png", dpi=180, bbox_inches="tight"); plt.show()
import base64
encoded=base64.b64encode(open("your_er_diagram.png","rb").read()).decode("utf-8")
penngrader2.submit("hw2", "test_er_diagram", encoded)
````
